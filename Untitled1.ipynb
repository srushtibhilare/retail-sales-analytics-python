```python
df.drop(columns=['Unnamed: 0'], inplace=True)
```

```python
df['promotion_intensity'] = df['promotion_intensity'].clip(lower=0)
```

```python
df['customer_sentiment'] = df['customer_sentiment'].clip(0, 1)
```

```python
df['effective_price'] = df['price'] - df['discount']
```

```python
df['price_diff_vs_competitor'] = df['price'] - df['competitor_price']
```

```python
df['demand_pressure'] = df['footfall'] / df['stock_level']
```

```python
df['ad_efficiency'] = df['footfall'] / df['ad_spend']
```

```python
import matplotlib.pyplot as plt

df[['price','discount','footfall','return_rate']].boxplot(figsize=(10,5))
plt.show()
```

```python
df = df[df['return_rate'] <= 0.20]
```

```python
df.to_csv("cleaned_retail_data.csv", index=False)
```

```python
import sqlite3

conn = sqlite3.connect("retail.db")
df.to_sql("retail_data", conn, if_exists="replace", index=False)
```

```python
query = """
SELECT discount_bucket, AVG(footfall) AS avg_footfall
FROM retail_data
GROUP BY discount_bucket;
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT *
FROM retail_data
WHERE return_rate > 0.15;
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT AVG(footfall / ad_spend) AS ad_efficiency
FROM retail_data;
"""
pd.read_sql(query, conn)
```
